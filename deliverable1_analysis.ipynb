```python
import numpy as np
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_excel("./Data/demographics.xls", engine="xlrd")
df = df.rename(columns={
    'Weight (kg)' : 'weight',
    'Height (meters)' : 'height'
})
df.describe()
df.info()
df["Group"] = df["Group"].replace({"CO": "Control", "PD": "Patient"})
df["Gender"] = df["Gender"].astype(str).str.strip().str.lower()
df.head()
```

```python
df['height'] = df['height'].fillna(df.groupby('Gender')['height'].transform('median'))
df['height_group'] = pd.cut(df['height'], bins=[1.4, 1.6, 1.75, 2.0])
df['weight'] = df['weight'].fillna(df.groupby(['Gender', 'height_group'])['weight'].transform('median'))
df
```
